# Session 2 — Text Representation

**Goal:** See how text can be represented in different ways.

Run the cells in order.

```python
texts = [
    'I like NLP',
    'I like machine learning',
    'NLP is interesting'
]

texts
```

## 1. Bag of Words (BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

## 2. TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray().round(2))
```

## 3. One-Hot Encoding

A simple example for individual words.

```python
words = ['cat', 'dog', 'bird']

for i, word in enumerate(words):
    vector = [0] * len(words)
    vector[i] = 1
    print(word, vector)
```

## 4. Word Embeddings (Word2Vec)

This tiny training example only demonstrates how to use Word2Vec. Similarity results are not meaningful with so little data.

Install Gensim if needed:

```python
%pip -q install gensim
```

```python
from gensim.models import Word2Vec

sentences = [
    ['i', 'like', 'nlp'],
    ['i', 'like', 'machine', 'learning'],
    ['nlp', 'is', 'interesting'],
    ['machine', 'learning', 'is', 'useful'],
    ['i', 'study', 'language']
]

model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, workers=1, epochs=100)

model.wv['nlp']
```

```python
model.wv.most_similar('nlp')
```

## 5. FastText

FastText also uses subword information. This toy example is only to observe the API.

```python
from gensim.models import FastText

fasttext_model = FastText(sentences, vector_size=10, window=2, min_count=1, workers=1, epochs=100)

fasttext_model.wv['nlp']
```

```python
fasttext_model.wv['nlpxyz']  # FastText can form vectors for unseen words using subword information
```

## 6. Sentence Embeddings

A pretrained model represents a whole sentence as a vector. Install the package if needed:

```python
%pip -q install sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

sentences_to_compare = [
    'I enjoy learning NLP.',
    'I like studying language technology.',
    'I dislike rainy weather.'
]

embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = embedding_model.encode(sentences_to_compare)

print(embeddings.shape)
```

## 7. Compare Sentence Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(embeddings)
print(similarity.round(2))
```
