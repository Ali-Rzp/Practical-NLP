# Practical NLP: Text Processing & Linguistic Analysis

A hands-on notebook demonstrating a basic NLP workflow using a shared text corpus.

**Goal:** Explore text preprocessing, tokenization, linguistic analysis, chunking, parsing, lexical relations, and regular expressions through executable Python examples.

## 1. Corpus

**Goal:** Create a small corpus and inspect its basic structure.

```python
corpus = [
    "Natural Language Processing helps computers understand human language!",
    "The students are studying NLP, machine learning, and artificial intelligence.",
    "A powerful model can analyze text, identify patterns, and extract information.",
    "The children were running quickly through the beautiful garden.",
    "Dr. Smith works at OpenAI. He studies language models and data science.",
    "I can't believe how useful text analysis can be; it's amazing."
]

for i, document in enumerate(corpus, start=1):
    print(f"{i}. {document}")

print("\\nNumber of documents:", len(corpus))
print("Total characters:", sum(len(doc) for doc in corpus))
```

## 2. Lowercasing

**Goal:** Convert text to lowercase and observe how capitalization changes.

```python
lowercased_corpus = [document.lower() for document in corpus]

for original, processed in zip(corpus, lowercased_corpus):
    print("Original:  ", original)
    print("Lowercase: ", processed)
    print()
```

## 3. Punctuation & Special Character Removal

**Goal:** Remove punctuation and selected non-alphanumeric characters while preserving word boundaries.

```python
import re
import string

text = "NLP is amazing!!! Let's analyze text: AI, ML & data-science."
without_punctuation = text.translate(str.maketrans("", "", string.punctuation))
without_special_chars = re.sub(r"[^A-Za-z0-9\s]", " ", text)

print("Original:                  ", text)
print("Without punctuation:       ", without_punctuation)
print("Without special characters:", without_special_chars)
```

## 4. Tokenization

**Goal:** Split text into sentences and word tokens.

```python
import nltk
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

from nltk.tokenize import sent_tokenize, word_tokenize

sample_text = "NLP is fascinating. It helps machines process human language!"
sentences = sent_tokenize(sample_text)
tokens = word_tokenize(sample_text)

print("Sentences:", sentences)
print("Word tokens:", tokens)
```

```python
tokenized_corpus = [word_tokenize(document) for document in corpus]
for i, tokens in enumerate(tokenized_corpus, start=1):
    print(f"Document {i}:", tokens)
```

## 5. Stop-Word Removal

**Goal:** Remove common words that may contribute less to a particular text-analysis task.

```python
nltk.download("stopwords", quiet=True)
from nltk.corpus import stopwords

english_stopwords = set(stopwords.words("english"))
sample_tokens = word_tokenize("This is a simple example of text processing with natural language processing.")
filtered_tokens = [token for token in sample_tokens if token.lower() not in english_stopwords]

print("Original tokens:", sample_tokens)
print("After stop-word removal:", filtered_tokens)
```

```python
filtered_corpus = [
    [token for token in word_tokenize(document) if token.lower() not in english_stopwords]
    for document in corpus
]
for i, tokens in enumerate(filtered_corpus, start=1):
    print(f"Document {i}:", tokens)
```

## 6. Stemming

**Goal:** Reduce related word forms to a common stem using a rule-based stemmer.

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["connect", "connected", "connecting", "connection", "study", "studies", "studying", "playing", "played", "player"]

for word in words:
    print(f"{word:15} -> {stemmer.stem(word)}")
```

## 7. Lemmatization

**Goal:** Convert inflected words to dictionary base forms (lemmas).

```python
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
words = ["cars", "children", "running", "better", "studies", "geese"]
for word in words:
    print(f"{word:12} -> {lemmatizer.lemmatize(word)}")
```

```python
# WordNet POS tags: n = noun, v = verb, a = adjective, r = adverb
examples = [("running", "v"), ("better", "a"), ("studies", "v"), ("cars", "n")]
for word, pos in examples:
    print(f"{word:12} ({pos}) -> {lemmatizer.lemmatize(word, pos=pos)}")
```

## 8. Part-of-Speech (POS) Tagging

**Goal:** Assign grammatical categories such as noun, verb, adjective, and adverb to tokens.

```python
nltk.download("averaged_perceptron_tagger_eng", quiet=True)
from nltk import pos_tag

pos_sentence = "The curious student quickly analyzed the linguistic data."
pos_tokens = word_tokenize(pos_sentence)
tagged_tokens = pos_tag(pos_tokens)
for token, tag in tagged_tokens:
    print(f"{token:15} {tag}")
```

## 9. Chunking

**Goal:** Group POS-tagged tokens into shallow syntactic phrases such as noun phrases (NP).

```python
from nltk import RegexpParser

chunk_grammar = r\"\"\"
NP: {<DT>?<JJ.*>*<NN.*>+}
VP: {<VB.*><NP>?}
\"\"\"
chunk_parser = RegexpParser(chunk_grammar)
chunk_sentence = "The young researcher developed a powerful language model."
chunk_tags = pos_tag(word_tokenize(chunk_sentence))
chunk_tree = chunk_parser.parse(chunk_tags)
print(chunk_tree)
```

```python
for subtree in chunk_tree.subtrees():
    if subtree.label() in {"NP", "VP"}:
        phrase = " ".join(token for token, tag in subtree.leaves())
        print(subtree.label(), "->", phrase)
```

## 10. Parsing

**Goal:** Represent hierarchical syntactic structure with a grammar and inspect a parse tree.

```python
from nltk import CFG, ChartParser

grammar = CFG.fromstring(
    "S -> NP VP\\n"
    "NP -> Det N | Det Adj N\\n"
    "VP -> V NP\\n"
    "Det -> 'the' | 'a'\\n"
    "Adj -> 'small' | 'young'\\n"
    "N -> 'robot' | 'scientist' | 'book'\\n"
    "V -> 'found' | 'read'"
)
parser = ChartParser(grammar)
parse_sentence = "the young scientist found a book".split()
for tree in parser.parse(parse_sentence):
    print(tree)
    tree.pretty_print()
```

## 11. Hypernyms and Hyponyms

**Goal:** Explore semantic relationships between concepts using WordNet.

- **Hypernym:** A broader category.
- **Hyponym:** A more specific member of a category.

```python
from nltk.corpus import wordnet as wn

synsets = wn.synsets("car", pos=wn.NOUN)
for synset in synsets[:3]:
    print("Synset:", synset.name())
    print("Definition:", synset.definition())
    print()
```

```python
car_synset = wn.synset("car.n.01")
print("Concept:", car_synset.name())

print("\\nHypernyms (broader concepts):")
for hypernym in car_synset.hypernyms():
    print("-", hypernym.name(), ":", hypernym.definition())

print("\\nHyponyms (more specific concepts):")
for hyponym in car_synset.hyponyms()[:10]:
    print("-", hyponym.name(), ":", hyponym.definition())
```

## 12. Regex Fundamentals

**Goal:** Use regular expressions to find and extract patterns from text.

```python
text = "Contact us at support@example.com or sales.team@company.org"
emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)
print("Extracted emails:", emails)
```

## 13. Regex: Digits, Numbers, and Dates

**Goal:** Match numeric patterns and date-like strings.

```python
text = "Order 4821 was placed on 2026-11-19. Total: 249.90 USD."
integers = re.findall(r"\b\d+\b", text)
decimals = re.findall(r"\b\d+\.\d{2}\b", text)
dates = re.findall(r"\b\d{4}-\d{2}-\d{2}\b", text)
print("Integers:", integers)
print("Decimal values:", decimals)
print("Dates:", dates)
```

## 14. Regex: Search, Match, and Find All

**Goal:** Compare common regex functions.

```python
text = "NLP2026 is the course code."
print("re.search:", re.search(r"\d+", text).group())
print("re.match:", re.match(r"NLP", text).group())
print("re.findall:", re.findall(r"[A-Z]+|\d+", text))
```

## 15. Regex: Substitution and Cleaning

**Goal:** Replace repeated whitespace, remove URLs, and normalize text patterns.

```python
text = "Visit https://example.com   for   more     information!"
without_urls = re.sub(r"https?://\S+", "", text)
normalized_spaces = re.sub(r"\s+", " ", without_urls).strip()
print("Original:", text)
print("Without URLs:", without_urls)
print("Normalized:", normalized_spaces)
```

## 16. Regex: Groups

**Goal:** Capture meaningful parts of a structured string.

```python
log_line = "2026-11-19 14:35:08 ERROR User login failed"
pattern = (r"(?P<date>\d{4}-\d{2}-\d{2}) "
           r"(?P<time>\d{2}:\d{2}:\d{2}) "
           r"(?P<level>[A-Z]+) (?P<message>.*)")
match = re.search(pattern, log_line)
if match:
    print("Date:", match.group("date"))
    print("Time:", match.group("time"))
    print("Level:", match.group("level"))
    print("Message:", match.group("message"))
```

## 17. Regex: Text Validation

**Goal:** Check whether strings follow simple expected patterns.

```python
samples = ["user_123", "Ali", "user-name", "user name", "data2026"]
pattern = r"^[A-Za-z0-9_]{3,15}$"
for sample in samples:
    result = bool(re.fullmatch(pattern, sample))
    print(f"{sample:15} -> {result}")
```

## 18. Mini Pipeline: Combine Preprocessing Steps

**Goal:** Apply cleaning, tokenization, stop-word removal, and lemmatization to one document.

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    tokens = word_tokenize(text)
    tokens = [token for token in tokens if token not in english_stopwords]
    tokens = [lemmatizer.lemmatize(token) for token in tokens]
    return tokens

sample_document = ("The students are studying Natural Language Processing, "
                   "and they are analyzing different documents!")
print("Original:", sample_document)
print("Processed tokens:", preprocess_text(sample_document))
```

## 19. Corpus-Level Comparison

**Goal:** Compare each original document with its processed representation.

```python
for index, document in enumerate(corpus, start=1):
    processed = preprocess_text(document)
    print(f"Document {index}")
    print("Original :", document)
    print("Processed:", processed)
    print("-" * 80)
```
